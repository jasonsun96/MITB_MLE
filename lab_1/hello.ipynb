```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

rng = np.random.default_rng(seed=42)
x = np.arange(0, 10).reshape(-1, 1)
y = 2 * x.ravel() + 1 + rng.normal(0, 0.5, size=10)

df = pd.DataFrame({'x': x.ravel(), 'y': y})
df
```

```python
model = LinearRegression().fit(x, y)
print(f'Slope:     {model.coef_[0]:.3f}  (true: 2)')
print(f'Intercept: {model.intercept_:.3f}  (true: 1)')

plt.scatter(x, y, label='data')
plt.plot(x, model.predict(x), color='red', label='fit')
plt.xlabel('x'); plt.ylabel('y'); plt.legend(); plt.title('Linear Regression')
plt.show()
```
